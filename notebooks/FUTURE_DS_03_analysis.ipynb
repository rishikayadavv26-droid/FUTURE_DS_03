# FUTURE_DS_03 – Marketing Funnel & Conversion Performance Analysis

```python
import pandas as pd
df=pd.read_csv('../data/marketing_funnel.csv')
df.head()
```

```python
g=df.groupby('Channel',as_index=False).sum(numeric_only=True)
g['Visitor_to_Lead']=g.Leads/g.Visitors
g['Lead_to_Opportunity']=g.Opportunities/g.Leads
g['Opportunity_to_Customer']=g.Customers/g.Opportunities
g['Overall_Conversion']=g.Customers/g.Visitors
g.sort_values('Overall_Conversion',ascending=False)
```

Improve the weakest funnel stage, strengthen lead nurturing, and evaluate channels using conversion and revenue rather than volume alone.
